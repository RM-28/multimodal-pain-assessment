```python
from pathlib import Path
from collections import Counter, defaultdict
import pandas as pd, numpy as np, kagglehub
import os

DATA_ROOT = Path(kagglehub.dataset_download("orvile/physiopain-dataset"))
all_files = [p for p in DATA_ROOT.rglob("*") if p.is_file()]

print(DATA_ROOT)
print("Total files:", len(all_files))
print("Extensions:", dict(Counter(p.suffix.lower() for p in all_files)))
```

```python
DATA_ROOT = kagglehub.dataset_download("orvile/physiopain-dataset")
print('Data root:', DATA_ROOT)

all_files = []
for folder, subfolders, filenames in os.walk(DATA_ROOT):
    for name in filenames:
        all_files.append(os.path.join(folder, name))

print('Total files:', len(all_files))
```

```python
extensions = {}
for path in all_files:
    ext = os.path.splitext(path)[1].lower()
    if ext in extensions:
        extensions[ext] = extensions[ext] + 1
    else:
        extensions[ext] = 1

print('File extensions:')
print(pd.Series(extensions).sort_values(ascending=False))
```

```python
folders = {}
for path in all_files:
    folder = os.path.relpath(os.path.dirname(path), DATA_ROOT)
    if folder in folders:
        folders[folder] = folders[folder] + 1
    else:
        folders[folder] = 1

for folder in sorted(folders):
    print(folders[folder], ' files  ', folder)
```

```python
eeg_root = os.path.join(DATA_ROOT, 'PhysioPain Dataset', 'Multimodal Pain Dataset', 'PROCESSED EEG DATA', '4Hz')

eeg_ids = {}
for folder in sorted(os.listdir(eeg_root)):
    ids = []
    for name in sorted(os.listdir(os.path.join(eeg_root, folder))):
        ids.append(os.path.splitext(name)[0])
    eeg_ids[folder] = ids
    print(folder, len(ids), ' files  ', ids[:5], ' ... ', ids[-5:])
```

```python
watch_root = os.path.join(DATA_ROOT, 'PhysioPain Dataset', 'Multimodal Pain Dataset', 'PROCESSED WATCH DATA')
watch_ids = {}
for folder in sorted(os.listdir(watch_root)):
    signal_folder = os.path.join(watch_root, folder, 'signal_4')
    if not os.path.exists(signal_folder):
        continue
    ids = []
    for name in sorted(os.listdir(signal_folder)):
        ids.append(os.path.splitext(name)[0])
    watch_ids[folder] = ids
    print(folder, len(ids), ' files  ', ids[:5], ' ... ', ids[-5:])
```

```python
folder = sorted(eeg_ids.keys())[0]
sample_eeg = os.path.join(eeg_root, folder, eeg_ids[folder][0] + '.csv')
print('Sample EEG file:', sample_eeg)

df_eeg = pd.read_csv(sample_eeg)
print('EEG data shape:', df_eeg.shape)
print('EEG data columns:', df_eeg.columns.tolist())
df_eeg.head()
```

```python
watch_folder = sorted(watch_ids.keys())[0]
sample_watch = os.path.join(watch_root, watch_folder, 'signal_4', watch_ids[watch_folder][0] + '.csv')

print('Sample WATCH file:', sample_watch)
df_watch = pd.read_csv(sample_watch)
print('WATCH data shape:', df_watch.shape)
print('WATCH data columns:', df_watch.columns.tolist())
df_watch.head()
```
